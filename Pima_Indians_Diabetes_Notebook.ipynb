# 1.0 Business problem
Healthcare organizations wants to predict whether a patient has diabetes based on their medical and diagnostic measurements.
So The aim of this project is to develop a machine-learning model that predicts whether a patient has diabetes based on their medical and diagnostic measurements.

**Target variable is:Outcome** `1 = Patient has diabetes`, `0 =Patient does not have diabetes `

This is a **binary classification** problem because the model predicts one of two classes: 

### 1.1 Business questions.
- Can we predict whether a patient is likely to have diabetes based on their medical measurements?
- Which patient characteristics are most strongly associated with diabetes?
- How accurately can the model identify patients who have diabetes?
- How well can the model identify patients who do not have diabetes?
- Can the model help healthcare professionals identify high-risk patients earlier?
- Which machine-learning model performs best for diabetes prediction?

### 1.2 Evaluation Metrics.
| Metric                   | What it tells us                                                   |
| ------------------------ | ------------------------------------------------------------------ |
| **Accuracy**             | Overall percentage of correct predictions                          |
| **Precision**            | Of patients predicted diabetic, how many actually have diabetes    |
| **Recall (Sensitivity)** | Of patients who actually have diabetes, how many the model detects |
| **F1-score**             | Balance between precision and recall                               |
| **Specificity**          | Ability to correctly identify patients without diabetes            |
| **Confusion Matrix**     | Shows TP, TN, FP and FN                                            |
| **ROC-AUC**              | Measures how well the model separates the two classes              |
